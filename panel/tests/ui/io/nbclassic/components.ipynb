```python
import pandas as pd
import panel as pn

pn.extension(
    'codeeditor', 'deckgl', 'echarts', 'filedropper', 'jsoneditor',
    'katex', 'perspective', 'tabulator', 'terminal', 'vega',
    comms='default', inline=False,
)
```

```python
button = pn.widgets.Button(name='Increment')
counter = pn.pane.Markdown('0', css_classes=['nbclassic-counter'])
button.on_click(lambda event: setattr(counter, 'object', str(int(counter.object) + 1)))
pn.Row(button, counter)
```

```python
data = pd.DataFrame({'x': [1, 2, 3], 'y': [4, 5, 6]})
pn.Column(
    pn.widgets.Tabulator(data, height=160),
    pn.pane.Perspective(data, height=200),
    pn.widgets.FileDropper(height=100),
    pn.widgets.CodeEditor(value='print("Panel")', height=100),
    pn.widgets.JSONEditor(value={'answer': 42}, height=160),
    pn.pane.ECharts({'xAxis': {'data': ['A', 'B']}, 'yAxis': {}, 'series': [{'type': 'bar', 'data': [1, 2]}]}, height=180),
    pn.pane.Vega({
        '$schema': 'https://vega.github.io/schema/vega-lite/v5.json',
        'mark': 'bar',
        'data': {
            'values': [{'x': 'A', 'y': 1}]
        },
        'encoding': {
            'x': {'field': 'x', 'type': 'nominal'},
            'y': {'field': 'y', 'type': 'quantitative'}
        }
    }, height=180),
    pn.pane.DeckGL({'initialViewState': {'longitude': 0, 'latitude': 0, 'zoom': 1}, 'layers': []}, height=180),
    pn.widgets.Terminal(height=100),
    pn.pane.LaTeX(r'E = mc^2'),
)
```
